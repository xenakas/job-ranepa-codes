

Базовые расчеты MRIO

Таблицы MRIO описывают глобальные межотраслевые потоки внутри и между странами для $k$ стран с матрицей транзакций $Z$ :
$$
Z=\left(\begin{array}{cccc}
Z_{1,1} & Z_{1,2} & \cdots & Z_{1, k} \\
Z_{2,1} & Z_{2,2} & \cdots & Z_{2, k} \\
\vdots & \vdots & \ddots & \vdots \\
Z_{k, 1} & Z_{k, 2} & \cdots & Z_{k, k}
\end{array}\right)
$$

Каждая подматрица на главной диагонали $\left(Z_{i, i}\right)$ представляет внутренние взаимодействия для каждой отрасли $n$. Внедиагональные матрицы $\left(Z_{i, j}\right)$ описывают торговлю из региона $i$ в регион $j$ (где $i, j=1, \ldots, k$ ) для каждой отрасли. Соответственно, глобальный конечный спрос может быть представлен формулой
$$
Y=\left(\begin{array}{cccc}
Y_{1,1} & Y_{1,2} & \cdots & Y_{1, k} \\
Y_{2,1} & Y_{2,2} & \cdots & Y_{2, k} \\
\vdots & \vdots & \ddots & \vdots \\
Y_{k, 1} & Y_{k, 2} & \cdots & Y_{k, k}
\end{array}\right)
$$
с конечным спросом, удовлетворяемым внутренним производством в главной диагонали $\left(Y_{i, i}\right)$ и прямым импортом конечного спроса из страны $i$ в $j$ через $Y_{i, j}$.

Таким образом, мировую экономику можно описать следующим образом:
$$
x=Z е+Y е
$$
где $e$ представляет собой вектор суммирования (вектор-столбец с единицами соответствующей размерности), а $x$ — общий объём промышленного производства.

Матрица прямых потребностей $A$ получается умножением $Z$ на диагонализированный и инвертированный выпуск отрасли $x$:
$$
A=Z \hat{x}^{-1}
$$

Основываясь на предположении о линейной экономике модели ввода-вывода и классическом моделировании спроса Леонтьева (см. Леонтьев 1970), общий объем промышленного производства $x$ может быть рассчитан для любого произвольного вектора конечного спроса $y$ путем умножения на матрицу общих потребностей. (Матрица Леонтьева) $L$.
$$
x=(\mathrm{I}-A)^{-1} y=L y
$$
где я определил как единичную матрицу размером $A$.

Глобальная многорегиональная система ввода-вывода может быть расширена за счет различных факторов производства $f_{h, i}$. Они могут представлять, среди прочего, добавленную стоимость, занятость и социальные факторы $(h$, где $h=1, \ldots, r$ ) по каждой стране. Векторы-строки факторов можно суммировать в матрице факторов производства $F$ :
$$
F=\left(\begin{array}{cccc}
f_{1,1} & f_{1,2} & \cdots & f_{1, k} \\
f_{2,1} & f_{2,2} & \cdots & f_{2, k} \\
\vdots & \vdots & \ddots & \vdots \\
f_{r, 1} & f_{r, 2} & \cdots & f_{r, k}
\end{array}\right)
$$
с коэффициентом производственного фактора $S$, определяемым выражением
$$
S=F \hat{x}^{-1}
$$

Если фактор производства оказывает необходимое воздействие на окружающую среду, оно также может иметь место на этапе конечного использования. В этом случае $G$ описывают воздействия, связанные с конечным спросом.

Счета, основанные на производстве (прямые территориальные требования) по каждой стране, представляют собой:
$$
D_{p b a}=F e+G e
$$

Множители для $F$ получаются следующим образом:
$$
М=С Л
$$

Общие потребности (следы в случае экологических требований) для любого заданного вектора конечного спроса $y$ определяются выражением
$$
D_{c b a}=M y
$$

Установление нулевого конечного спроса, удовлетворяемого внутри страны, $Y_{i, i}$ $\left(Y_t=Y-Y_{i, j} \mid i=j\right)$ позволяет рассчитать фактор производства, происходящий за границей (воплощенный в импорте)
$$
D_{i m p}=S M Y_t
$$

Факторы производства, происходящие внутри страны для удовлетворения конечного спроса в других странах, определяются следующим образом:
$$
D_{\text {exp }}=\widehat{S M Y_t e}
$$

Общую потребность для каждой страны можно получить путем суммирования секторов для каждого счета $\left(D_{c b a}, D_{i m p}\right.$ и $D_{\text {exp }}$ ). В случае $D_{c b a}$ необходимо добавить любые воздействия, связанные с использованием $(G)$. Используя этот подход, следы для каждой страны $i$ удовлетворяют следующему:
$$
D_{c b a}^i=D_{p b a}^i+D_{i m p}^i-D_{\text {exp }}^i
$$

Агрегация


Для агрегации системы MRIO матрица $S_k$ определяет матрицу агрегации для регионов, а $S_n$ — матрицу агрегации для секторов.
$$
S_k=\left(\begin{array}{cccc}
b_{1,1} & b_{1,2} & \cdots & b_{1, k} \\
b_{2,1} & b_{2,2} & \cdots & b_{2, k} \\
\vdots & \vdots & \ddots & \vdots \\
b_{w, 1} & b_{w, 2} & \cdots & b_{w, k}
\end{array}\right) S_n=\left(\begin{array}{cccc}
b_{1,1} & b_{1,2} & \cdots & b_{1, n} \\
b_{2,1} & b_{2,2} & \cdots & b_{2, n} \\
\vdots & \vdots & \ddots & \vdots \\
b_{x, 1} & b_{x, 2} & \cdots & b_{x, n}
\end{array}\right)
$$

При этом $w$ и $x$ определяют совокупное количество стран и секторов соответственно. Записям $b$ присваивается значение 1, если сектор/страна столбца принадлежит агрегированному сектору/региону в соответствующей строке, и н\оль в противном случае. Полная матрица агрегирования $S$ задается произведением Кронекера $\otimes$ $S_k$ и $S_n:$
$$
S=S_k \otimes S_n
$$

Агрегированную систему ввода-вывода можно получить с помощью
$$
Z_{a g g}=S Z S^{\mathrm{T}}
$$
и
$$
Y_{a g g}=S Y\left(S_k \otimes \mathrm{I}\right)^{\mathrm{T}}
$$
причем я определил как матрицу тождественности размера категорий конечного спроса на страну.

Факторы производства агрегируются по
$$
F_{a g g}=F S^{\mathrm{T}}
$$
и влияние на конечный спрос
$$
G_{a g g}=G\left(S_k \otimes \mathrm{I}\right)^{\mathrm{T}}
$$
